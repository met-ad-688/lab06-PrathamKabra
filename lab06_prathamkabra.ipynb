# Step 1: Load the Dataset {.unnumbered}


```python
import pandas as pd
import plotly.express as px
import plotly.io as pio
pio.renderers.default = "vscode"
from pyspark.sql import SparkSession
from pyspark.sql.functions import col


# Initialize Spark Session
spark = SparkSession.builder.appName("LightcastData").getOrCreate()

# Load Data
df = spark.read.option("header", "true").option("inferSchema", "true").option("multiLine","true").option("escape", "\"").csv("./data/lightcast_job_postings.csv")

# Show Schema and Sample Data
#df.printSchema()
#df.show(5)
```

# Salary Distribution by Employment Type
- Identify salary trends across different employment types.
- **Filter the dataset**
  - Remove records where **salary is missing or zero**.
- **Aggregate Data**
  - Group by **employment type** and compute salary distribution.
- **Visualize results**
  - Create a **box plot** where:
    - **X-axis** = `EMPLOYMENT_TYPE_NAME`
    - **Y-axis** = `SALARY_FROM`
  - Customize **colors, fonts, and styles** to avoid a **2.5-point deduction**.
- **Explanation:** Write two sentences about what the graph reveals.



```python
import plotly.express as px

# 1. Filter out missing or zero salaries
pdf = (
    df
    .filter(
        (df["SALARY_FROM"].isNotNull()) &
        (df["SALARY_FROM"] > 0)
    )
    .select("EMPLOYMENT_TYPE_NAME", "SALARY_FROM")
    .toPandas()
)

# 2. Box-plot of salary distributions by employment type
fig = px.box(
    pdf,
    x="EMPLOYMENT_TYPE_NAME",
    y="SALARY_FROM",
    title="Salary Distribution by Employment Type",
    color_discrete_sequence=["#636EFA"]  # single brand color
)

# 3. Tweak fonts & layout to dodge that 2.5-point hit
fig.update_layout(
    font_family="Arial",
    title_font_size=16,
    xaxis_title="Employment Type",
    yaxis_title="Salary From",
    margin=dict(l=40, r=40, t=60, b=40)
)

fig.show()
```

# Job Posting Trends Over Time
- Analyze how job postings fluctuate over time.
- **Aggregate Data**
  - Count job postings per **posted date (`POSTED`)**.
- **Visualize results**
  - Create a **line chart** where:
    - **X-axis** = `POSTED`
    - **Y-axis** = `Number of Job Postings`
  - Apply custom colors and font styles.
- **Explanation:** Write two sentences about what the graph reveals.

```python
# Code for 3rd question here
import plotly.express as px

# 1. Aggregate job postings by date
pdf = (
    df
    .select("POSTED")
    .groupBy("POSTED")
    .count()
    .withColumnRenamed("count", "Num_Postings")
    .orderBy("POSTED")
    .toPandas()
)

# 2. Build the line chart
fig = px.line(
    pdf,
    x="POSTED",
    y="Num_Postings",
    title="Job Posting Trends Over Time",
    color_discrete_sequence=["#636EFA"]
)

# 3. Tweak fonts & layout
fig.update_layout(
    font_family="Arial",
    title_font_size=16,
    xaxis_title="Date Posted",
    yaxis_title="Number of Job Postings",
    margin=dict(l=40, r=40, t=60, b=40)
)

fig.show()
```

# Top 10 Job Titles by Count
- Identify the most frequently posted job titles.
- **Aggregate Data**
  - Count the occurrences of each **job title (`TITLE_NAME`)**.
  - Select the **top 10 most frequent titles**.
- **Visualize results**
  - Create a **bar chart** where:
    - **X-axis** = `TITLE_NAME`
    - **Y-axis** = `Job Count`
  - Apply custom colors and font styles.
- **Explanation:** Write two sentences about what the graph reveals.

```python
# Code for 4th question here
import plotly.express as px
from pyspark.sql import functions as F


# 1. Aggregate and get top 10 titles by count
pdf = (
    df
    .groupBy("TITLE_NAME")
    .count()
    .withColumnRenamed("count", "Job_Count")
    .orderBy(F.desc("Job_Count"))
    .limit(10)
    .toPandas()
)

# 2. Build the bar chart
fig = px.bar(
    pdf,
    x="TITLE_NAME",
    y="Job_Count",
    title="Top 10 Job Titles by Count",
    color_discrete_sequence=["#636EFA"]
)

# 3. Tweak fonts & layout
fig.update_layout(
    font_family="Arial",
    title_font_size=16,
    xaxis_title="Job Title",
    yaxis_title="Number of Postings",
    xaxis_tickangle=-45,
    margin=dict(l=40, r=40, t=60, b=100)
)

fig.show()
```

# Remote vs On-Site Job Postings
- Compare the proportion of remote and on-site job postings.
- **Aggregate Data**
  - Count job postings by **remote type (`REMOTE_TYPE_NAME`)**.
- **Visualize results**
  - Create a **pie chart** where:
    - **Labels** = `REMOTE_TYPE_NAME`
    - **Values** = `Job Count`
  - Apply custom colors and font styles.
- **Explanation:** Write two sentences about what the graph reveals.

```python
# Code for 5th question here
import plotly.express as px

# 1. Aggregate counts by remote type
pdf = (
    df
    .groupBy("REMOTE_TYPE_NAME")
    .count()
    .withColumnRenamed("count", "Job_Count")
    .toPandas()
)

# 2. Build the pie chart
fig = px.pie(
    pdf,
    names="REMOTE_TYPE_NAME",
    values="Job_Count",
    title="Remote vs On-Site Job Postings",
    color_discrete_sequence=["#636EFA", "#EF553B"]  # two-tone palette
)

# 3. Tweak fonts & layout
fig.update_traces(textinfo="percent+label", 
                  marker=dict(line=dict(color="white", width=1)))
fig.update_layout(
    font_family="Arial",
    title_font_size=16,
    margin=dict(l=40, r=40, t=60, b=40)
)

fig.show()
```

# Skill Demand Analysis by Industry (Stacked Bar Chart)
- Identify which skills are most in demand in various industries.
- **Aggregate Data**
  - Extract **skills** from job postings.
  - Count occurrences of skills grouped by **NAICS industry codes**.
- **Visualize results**
  - Create a **stacked bar chart** where:
    - **X-axis** = `Industry`
    - **Y-axis** = `Skill Count`
    - **Color** = `Skill`
  - Apply custom colors and font styles.
- **Explanation:** Write two sentences about what the graph reveals.

```python
# Code for 6th question here
from pyspark.sql import functions as F
import plotly.express as px

# 1. Extract and explode skills (assuming comma-separated in a column named 'SKILLS')
df_skills = (
    df
    .select("NAICS2_NAME", "SKILLS")
    .filter(F.col("SKILLS").isNotNull())
    .withColumn("Skill", F.explode(F.split(F.col("SKILLS"), ",")))
    .withColumn("Skill", F.trim(F.col("Skill")))
)

# 2. Count occurrences of each skill per industry
skill_counts = (
    df_skills
    .groupBy("NAICS2_NAME", "Skill")
    .count()
    .withColumnRenamed("count", "Skill_Count")
)

# 3. Move to pandas for plotting
pdf = skill_counts.toPandas()

# 4. Build the stacked bar chart
fig = px.bar(
    pdf,
    x="NAICS2_NAME",
    y="Skill_Count",
    color="Skill",
    title="Skill Demand by Industry",
    labels={"NAICS2_NAME": "Industry", "Skill_Count": "Skill Count"}
)

# 5. Tweak fonts, legend, and layout
fig.update_layout(
    barmode="stack",
    font_family="Arial",
    title_font_size=16,
    legend_title_text="Skill",
    xaxis_title="Industry",
    yaxis_title="Total Skill Mentions",
    margin=dict(l=40, r=40, t=60, b=100)
)

fig.show()
```


# Salary Analysis by ONET Occupation Type (Bubble Chart)
- Analyze how salaries differ across ONET occupation types.
- **Aggregate Data**
  - Compute **median salary** for each occupation in the **ONET taxonomy**.
- **Visualize results**
  - Create a **bubble chart** where:
    - **X-axis** = `ONET_NAME`
    - **Y-axis** = `Median Salary`
    - **Size** = Number of job postings
  - Apply custom colors and font styles.
- **Explanation:** Write two sentences about what the graph reveals.

```python
# Code for 7th question here
from pyspark.sql import functions as F
import plotly.express as px

# 1. Aggregate data: median salary and count of postings per ONET occupation
agg = (
    df
    .filter(df["SALARY_FROM"] > 0)
    .groupBy("ONET_NAME")
    .agg(
        F.expr("percentile_approx(SALARY_FROM, 0.5)").alias("Median_Salary"),
        F.count("*").alias("Num_Postings")
    )
)

pdf = agg.toPandas()

# 2. Build the bubble chart
fig = px.scatter(
    pdf,
    x="ONET_NAME",
    y="Median_Salary",
    size="Num_Postings",
    title="Salary Analysis by ONET Occupation Type",
    size_max=60,
    color_discrete_sequence=["#636EFA"],
    labels={
        "ONET_NAME": "Occupation",
        "Median_Salary": "Median Salary",
        "Num_Postings": "Number of Postings"
    }
)

# 3. Tweak fonts & layout
fig.update_layout(
    font_family="Arial",
    title_font_size=16,
    xaxis_tickangle=-45,
    margin=dict(l=40, r=40, t=60, b=120)
)

fig.show()
```
